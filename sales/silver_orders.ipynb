```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
df = spark.read.format("parquet").load("abfss://sales@adls7786.dfs.core.windows.net/orders/bronze")
df = df.withColumn("order_date",F.to_timestamp(F.col("order_date")))
df = df.withColumn("year",F.year(F.col("order_date")))
# df = df.withColumn("flag",F.dense_rank().over(Window.partitionBy(F.col("year")).orderBy(F.col("total_amount").desc()))).withColumn("flag_rank",F.rank().over(Window.partitionBy(F.col("year")).orderBy(F.col("total_amount").desc()))).withColumn("flag_number",F.row_number().over(Window.partitionBy(F.col("year")).orderBy(F.col("total_amount").desc())))
# df.display()

df.write.format("delta").mode("append").option("path","abfss://sales@adls7786.dfs.core.windows.net/orders/silver").saveAsTable("sales.silver.orders")
```

```python
class windows:

    def dense_rank(self,df):
        df_dense_rank = df.withColumn("dense_rank",F.dense_rank().over(Window.partitionBy(F.col("year")).orderBy(F.col("total_amount").desc())))
        return df_dense_rank

    def rank(self,df):
        df_rank = df.withColumn("rank",F.rank().over(Window.partitionBy(F.col("year")).orderBy(F.col("total_amount").desc())))
        return df_rank

    def row_number(self,df):
        df_row_number = df.withColumn("row_number",F.row_number().over(Window.partitionBy(F.col("year")).orderBy(F.col("total_amount").desc())))
        return df_row_number
```
