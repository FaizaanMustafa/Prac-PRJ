```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
dbutils.widgets.text("source_folder","")
dbutils.widgets.text("source_sub_folder","")
dbutils.widgets.text("target_folder","")
dbutils.widgets.text("target_sub_folder","")

source_folder = dbutils.widgets.get("source_folder")
target_folder = dbutils.widgets.get("target_folder")
source_sub_folder = dbutils.widgets.get("source_sub_folder")
target_sub_folder = dbutils.widgets.get("target_sub_folder")
```

```python
df = spark.readStream.format("cloudFiles").option("cloudFiles.format","parquet") \
            .option("cloudFiles.schemaLocation",f"abfss://sales@adls7786.dfs.core.windows.net/{source_folder}/checkpoint_{source_folder}").load(f"abfss://sales@adls7786.dfs.core.windows.net/{source_folder}/{source_sub_folder}")

df.writeStream.format("parquet").outputMode("append").option("checkpointLocation",f"abfss://sales@adls7786.dfs.core.windows.net/{target_folder}/checkpoint_{target_folder}").option("path",f"abfss://sales@adls7786.dfs.core.windows.net/{target_folder}/{target_sub_folder}").trigger(once = True).start()
```
