```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
df = spark.read.format("parquet").load("abfss://sales@adls7786.dfs.core.windows.net/customers/bronze")

df = df.withColumn("domain",F.split(F.col("email"),"@")[1]).withColumn("full_name",F.concat_ws(" ",F.col("first_name"),F.col("last_name"))).drop("first_name","last_name")

df.write.format("delta").mode("overwrite").option("path","abfss://sales@adls7786.dfs.core.windows.net/customers/silver").saveAsTable("sales.silver.customers")
```
