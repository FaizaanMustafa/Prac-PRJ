```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = spark.read.format("parquet").load("abfss://sales@adls7786.dfs.core.windows.net/products/bronze")

```

```python
df.display()
```

```python
df.createOrReplaceTempView("product")


```

```python
%sql

create or replace function sales.silver.silver_discount(p_price double)
returns double
language sql
return p_price *0.90
```

```python
%sql

select product_id,price,round(sales.silver.silver_discount(price),2) as discounted_price from product;
```

```python
df.write.format("delta").mode("append").option("path","abfss://sales@adls7786.dfs.core.windows.net/products/silver").saveAsTable("sales.silver.products")
```
