```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
dbutils.widgets.text("source_folder", "")
dbutils.widgets.text("source_direct", "")
dbutils.widgets.text("target_folder", "")
dbutils.widgets.text("target_direct", "")
source_folder = dbutils.widgets.get("source_folder")
source_direct = dbutils.widgets.get("source_direct")
target_folder = dbutils.widgets.get("target_folder")
target_direct = dbutils.widgets.get("target_direct")
```

```python
try:
    df = spark.read.format("csv").option("header","true").option("inferSchema","true").load(f"abfss://olympics@adls7786.dfs.core.windows.net/{source_folder}/{source_direct}")
    df.write.format("delta").mode("overwrite").option("path",f"abfss://olympics@adls7786.dfs.core.windows.net/{target_folder}/{target_direct}").saveAsTable(f"olympics.bronze.{target_direct}")
except Exception as e:
    print(f"error as {e}")
```
