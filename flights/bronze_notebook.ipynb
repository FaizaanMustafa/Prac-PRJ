```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
dbutils.widgets.text("source_folder","")
dbutils.widgets.text("source_sub_folder","")
dbutils.widgets.text("target_sub_folder","")
source_folder = dbutils.widgets.get("source_folder")
source_sub_folder = dbutils.widgets.get("source_sub_folder")
target_sub_folder = dbutils.widgets.get("target_sub_folder")

```

```python
df = spark.readStream.format("cloudFiles").option("cloudFiles.format","csv").option("cloudFiles.schemaLocation",f"abfss://flights@adls7786.dfs.core.windows.net/{source_folder}/checkpoint_{source_folder}").load(f"abfss://flights@adls7786.dfs.core.windows.net/{source_folder}/{source_sub_folder}")
df.writeStream.format("delta").outputMode("append").trigger(once=True).option("checkpointLocation",f"abfss://flights@adls7786.dfs.core.windows.net/{source_folder}/checkpoint_{source_folder}").option("path",f"abfss://flights@adls7786.dfs.core.windows.net/{source_folder}/{target_sub_folder}").toTable(f"flights.bronze.{source_folder}")
```
