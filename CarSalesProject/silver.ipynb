```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
df = spark.read.format("csv").option("header","true").option("inferSchema","true").load("abfss://carsales@adls7786.dfs.core.windows.net/raw")
df.display()
```

```python
df = df.withColumn("model_category",F.split(F.col("Model_ID"),'-')[0]) \
    .withColumn("revnue_per_unit",F.col('Revenue')/F.col('Units_Sold'))
df.write.format("parquet").mode("overwrite").option("path","abfss://carsales@adls7786.dfs.core.windows.net/silver").save()
```
