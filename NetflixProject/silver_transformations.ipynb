```python
from pyspark.sql import functions as F

dbutils.widgets.text("source_folder","")
dbutils.widgets.text("target_folder","")
source_folder = dbutils.widgets.get("source_folder")
target_folder = dbutils.widgets.get("target_folder")
```

```python
try:
    df = spark.read.format("csv").option("header","true").option("inferSchema","true").load(f"abfss://netflix@adls7786.dfs.core.windows.net/raw/{source_folder}")

    df.write.format("delta").mode("overwrite").option("path",f"abfss://netflix@adls7786.dfs.core.windows.net/bronze/{target_folder}").saveAsTable(f"netflix.bronze.{target_folder}")
except Exception as e:
    print(f"error:{e}")
```
